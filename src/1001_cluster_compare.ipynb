```python
import pandas as pd

df = pd.read_csv('/home/ys0660/2507Sub/textclustering/0915/Topicker/src/2nd_relabel.csv')
```

```python
df['1st_relabel'] = df['1st_relabel'].fillna(df['cluster_label'])
df['2nd_relabel'] = df['2nd_relabel'].fillna(df['cluster_label'])
```

```python
import matplotlib.pyplot as plt

fig, axes = plt.subplots(1, 3, figsize=(15,4))

df['label_id'].value_counts().sort_index().plot.bar(ax=axes[0], title="Truth (label_id)")
df['1st_relabel'].value_counts().sort_index().plot.bar(ax=axes[1], title="1st Relabel")
df['2nd_relabel'].value_counts().sort_index().plot.bar(ax=axes[2], title="2nd Relabel")

plt.show()
```

```python
import seaborn as sns
import pandas as pd

cm1 = pd.crosstab(df['label_id'], df['1st_relabel'])
cm2 = pd.crosstab(df['label_id'], df['2nd_relabel'])

plt.figure(figsize=(14,6))

plt.subplot(1,2,1)
sns.heatmap(cm1, annot=True, fmt="d", cmap="Blues")
plt.title("Truth vs 1st Relabel")
plt.xlabel("1st_relabel")
plt.ylabel("label_id")

plt.subplot(1,2,2)
sns.heatmap(cm2, annot=True, fmt="d", cmap="Greens")
plt.title("Truth vs 2nd Relabel")
plt.xlabel("2nd_relabel")
plt.ylabel("label_id")

plt.tight_layout()
plt.show()
```

```python
from sklearn.metrics import accuracy_score, f1_score

y_true = df['label_id']
y_pred1 = df['1st_relabel']
y_pred2 = df['2nd_relabel']

print("1st relabel accuracy:", accuracy_score(y_true, y_pred1))
print("2nd relabel accuracy:", accuracy_score(y_true, y_pred2))

print("1st relabel F1(macro):", f1_score(y_true, y_pred1, average='macro'))
print("2nd relabel F1(macro):", f1_score(y_true, y_pred2, average='macro'))
```
