```python
import pandas as pd

# CSV 읽기
df = pd.read_csv('/home/ys0660/2507Sub/textclustering/0915/Topicker/src/2nd_relabel.csv')

# 규칙 적용 함수
def compare_cols(row):
    a, b, c = row['cluster_label'], row['1st_relabel'], row['2nd_relabel']
    if a == b == c:
        return 'clean'
    elif len({a, b, c}) == 3:
        return 'dirty'
    else:
        return ''  # 미표기 (빈칸)

# 새 컬럼 생성
df['dirty'] = df.apply(compare_cols, axis=1)

df['all_okay'] = df.apply(
    lambda row: 1 if row['dirty'] == 'clean' and row['1st_relabel'] == row['label_id'] else 0,
    axis=1
)

# 엑셀로 저장
df.to_csv('/home/ys0660/2507Sub/textclustering/0915/Topicker/result.csv', index=False)

```
